<h2 style="color:DodgerBlue">Описание проекта:</h2>

Описание задачи:
Создать базовый класс Supplier в C#, который будет представлять информацию о
поставщиках товаров или услуг. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу Supplier:
• Атрибуты: ID поставщика (SupplierId), Название компании (CompanyName),
Тип продукции (ProductType).
• Методы:
o GetCompanyInfo(): метод для получения информации о компании.
o ProvideQuote(): метод для предоставления котировки на товары или
услуги.
o SubmitOrder(): метод для отправки заказа поставщику.
Требования к производным классам:
1. Производитель (Manufacturer): Должен содержать дополнительные атрибуты,
такие как Год основания (FoundedYear). Метод ProvideQuote() должен быть
переопределен для включения информации о годе основания компании в
котировку.
2. Ритейлер (Retailer): Должен содержать дополнительные атрибуты, такие как
Расположение магазина (StoreLocation). Метод SubmitOrder() должен быть
переопределен для добавления информации о расположении магазина при
отправке заказа.
3. Импортер (Importer) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Страна происхождения товара
(OriginCountry). Метод GetCompanyInfo() должен быть переопределен для
отображения страны происхождения товара вместе с остальной
информацией о компании.

```
using System;
using System.Collections.Generic;

// ===== Интерфейсы =====

public interface IInternational
{
    string Country { get; set; }
    void ShowInternationalStatus();
}

public interface IDiscountProvider
{
    double DiscountRate { get; set; }
    void ApplyDiscount(double percent);
}

// ===== Базовый класс Supplier =====

public class Supplier
{
    private int _supplierId;
    private string _companyName;
    private string _productType;

    public int SupplierId { get => _supplierId; set => _supplierId = value; }
    public string CompanyName { get => _companyName; set => _companyName = value; }
    public string ProductType { get => _productType; set => _productType = value; }

    // Дополнительные свойства
    public string Country { get; set; }
    public double Rating { get; set; }
    public string Email { get; set; }
    public bool IsActive { get; set; }

    // Конструктор
    public Supplier(int supplierId, string companyName, string productType)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;
        Country = "Россия";
        Rating = 4.5;
        Email = $"{companyName.Replace(" ", "").ToLower()}@mail.com";
        IsActive = true;
    }

    // ===== Методы =====

    public virtual void GetCompanyInfo()
    {
        Console.WriteLine($"ID поставщика: {SupplierId}");
        Console.WriteLine($"Название компании: {CompanyName}");
        Console.WriteLine($"Тип продукции: {ProductType}");
        Console.WriteLine($"Страна: {Country}, Рейтинг: {Rating}");
        Console.WriteLine($"Email: {Email}, Активен: {IsActive}");
    }

    // Полиморфизм: перегрузка метода
    public void ProvideQuote()
    {
        Console.WriteLine($"Котировка от {CompanyName} на товары типа: {ProductType}");
    }

    public void ProvideQuote(double discount)
    {
        Console.WriteLine($"Котировка от {CompanyName} на {ProductType} со скидкой {discount}%");
    }

    // Виртуальные методы для переопределения
    public virtual void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен поставщику: {CompanyName}");
    }

    public virtual void CollaborateWith(Supplier partner)
    {
        Console.WriteLine($"{CompanyName} начинает сотрудничество с {partner.CompanyName}");
        Console.WriteLine($"Направление: {ProductType} + {partner.ProductType}");
    }

    public void Deactivate()
    {
        IsActive = false;
        Console.WriteLine($"{CompanyName} временно приостановила деятельность.");
    }

    public void SendEmail(string message)
    {
        Console.WriteLine($"Отправка письма на {Email}: '{message}'");
    }

    public void UpdateRating(double newRating)
    {
        Rating = newRating;
        Console.WriteLine($"Рейтинг компании {CompanyName} обновлен до {Rating}");
    }
}

// ===== Производный класс Manufacturer =====

public class Manufacturer : Supplier
{
    public int FoundedYear { get; set; }
    public int ProductionCapacity { get; set; }
    public bool HasISO9001 { get; set; }
    public string MainProduct { get; set; }
    public int EmployeesCount { get; set; }

    public Manufacturer(int supplierId, string companyName, string productType, int foundedYear)
        : base(supplierId, companyName, productType)
    {
        FoundedYear = foundedYear;
        ProductionCapacity = 10000;
        HasISO9001 = true;
        MainProduct = productType;
        EmployeesCount = 300;
    }

    // Перекрытие (override)
    public override void SubmitOrder()
    {
        Console.WriteLine($"Производитель {CompanyName} принял заказ на {ProductType}");
    }

    // Переопределение базового метода
    public override void CollaborateWith(Supplier partner)
    {
        Console.WriteLine($"{CompanyName} (производитель) запускает проект с {partner.CompanyName}");
    }

    // Новые методы
    public void HireEmployees(int count)
    {
        EmployeesCount += count;
        Console.WriteLine($"{CompanyName} наняла {count} новых сотрудников. Всего: {EmployeesCount}");
    }

    public void SupplyToRetailer(Retailer retailer)
    {
        Console.WriteLine($"{CompanyName} поставляет товары ритейлеру {retailer.CompanyName} ({retailer.StoreLocation})");
    }

    public void UpgradeCapacity(int newCapacity)
    {
        ProductionCapacity = newCapacity;
        Console.WriteLine($"{CompanyName} увеличивает мощность до {ProductionCapacity} единиц/мес.");
    }
}

// ===== Производный класс Retailer =====

public class Retailer : Supplier, IDiscountProvider
{
    public string StoreLocation { get; set; }
    public int BranchCount { get; set; }
    public string ManagerName { get; set; }
    public double AverageMonthlySales { get; set; }

    public double DiscountRate { get; set; }

    public Retailer(int supplierId, string companyName, string productType, string storeLocation)
        : base(supplierId, companyName, productType)
    {
        StoreLocation = storeLocation;
        BranchCount = 5;
        ManagerName = "Иванов И.И.";
        AverageMonthlySales = 1500000;
        DiscountRate = 5.0;
    }

    public override void SubmitOrder()
    {
        Console.WriteLine($"Заказ отправлен ритейлеру {CompanyName}");
        Console.WriteLine($"Местоположение: {StoreLocation}");
        Console.WriteLine($"Менеджер: {ManagerName}");
    }

    public void OrderFromManufacturer(Manufacturer manufacturer, int quantity)
    {
        Console.WriteLine($"{CompanyName} заказывает {quantity} единиц у {manufacturer.CompanyName}");
        manufacturer.ProvideQuote();
    }

    // Перегрузка метода (Overload)
    public void ApplyDiscount()
    {
        Console.WriteLine($"{CompanyName} применяет стандартную скидку {DiscountRate}% на {ProductType}");
    }

    public void ApplyDiscount(double percent)
    {
        DiscountRate = percent;
        Console.WriteLine($"{CompanyName} применяет скидку {DiscountRate}% на продукцию {ProductType}");
    }

    public void OpenNewBranch(string address)
    {
        BranchCount++;
        Console.WriteLine($"{CompanyName} открыла новый филиал по адресу: {address}. Всего филиалов: {BranchCount}");
    }
}

// ===== Производный класс Importer =====

public class Importer : Supplier, IInternational, IDiscountProvider
{
    public string OriginCountry { get; set; }
    public string LicenseNumber { get; set; }
    public double CustomsFee { get; set; }
    public string TransportType { get; set; }

    public double DiscountRate { get; set; }
    public string Country { get; set; }

    public Importer(int supplierId, string companyName, string productType, string originCountry)
        : base(supplierId, companyName, productType)
    {
        OriginCountry = originCountry;
        LicenseNumber = $"LIC-{supplierId}-X";
        CustomsFee = 12.5;
        TransportType = "Морской транспорт";
        DiscountRate = 0;
        Country = originCountry;
    }

    public override void GetCompanyInfo()
    {
        base.GetCompanyInfo();
        Console.WriteLine($"Импорт из: {OriginCountry}, Лицензия: {LicenseNumber}");
        Console.WriteLine($"Таможенная пошлина: {CustomsFee}%, Транспорт: {TransportType}");
    }

    public void ImportForSupplier(Supplier supplier)
    {
        Console.WriteLine($"{CompanyName} импортирует товары для {supplier.CompanyName}");
    }

    public void ShowInternationalStatus()
    {
        Console.WriteLine($"{CompanyName} — международный поставщик из {OriginCountry}");
    }

    public void ApplyDiscount(double percent)
    {
        DiscountRate = percent;
        Console.WriteLine($"{CompanyName} предоставляет скидку {DiscountRate}% на импортируемые товары.");
    }
}

// ===== Обобщённый класс (Generic) =====

public class SupplierCollection<T> where T : Supplier
{
    private List<T> _items = new List<T>();

    public void Add(T item) => _items.Add(item);
    public void Remove(T item) => _items.Remove(item);

    public void DisplayAll()
    {
        Console.WriteLine("\n=== Список поставщиков ===");
        foreach (var item in _items)
        {
            item.GetCompanyInfo();
            Console.WriteLine("--------------------------");
        }
    }
}

Supplier supplier1 = new Supplier(1, "ООО 'Поставщик+'", "Электроника");
Manufacturer manufacturer1 = new Manufacturer(2, "Завод 'Электроникс'", "Компьютеры", 1995);
Retailer retailer1 = new Retailer(3, "Сеть 'ТехноМир'", "Бытовая техника", "ул. Центральная, 123");
Importer importer1 = new Importer(4, "ИмпортТрейд", "Мебель", "Италия");

supplier1.ProvideQuote();
supplier1.ProvideQuote(15);

manufacturer1.SubmitOrder();
manufacturer1.HireEmployees(25);

retailer1.SubmitOrder();
retailer1.ApplyDiscount();
retailer1.ApplyDiscount(10);

importer1.GetCompanyInfo();
importer1.ShowInternationalStatus();
importer1.ApplyDiscount(7.5);

importer1.ImportForSupplier(supplier1);
manufacturer1.SupplyToRetailer(retailer1);
retailer1.OrderFromManufacturer(manufacturer1, 50);

SupplierCollection<Supplier> collection = new SupplierCollection<Supplier>();
collection.Add(supplier1);
collection.Add(manufacturer1);
collection.Add(retailer1);
collection.Add(importer1);
collection.DisplayAll();

```
