<h2 style="color:DodgerBlue">Описание проекта:</h2>

Описание задачи:
Создать базовый класс Supplier в C#, который будет представлять информацию о
поставщиках товаров или услуг. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу Supplier:
• Атрибуты: ID поставщика (SupplierId), Название компании (CompanyName),
Тип продукции (ProductType).
• Методы:
o GetCompanyInfo(): метод для получения информации о компании.
o ProvideQuote(): метод для предоставления котировки на товары или
услуги.
o SubmitOrder(): метод для отправки заказа поставщику.
Требования к производным классам:
1. Производитель (Manufacturer): Должен содержать дополнительные атрибуты,
такие как Год основания (FoundedYear). Метод ProvideQuote() должен быть
переопределен для включения информации о годе основания компании в
котировку.
2. Ритейлер (Retailer): Должен содержать дополнительные атрибуты, такие как
Расположение магазина (StoreLocation). Метод SubmitOrder() должен быть
переопределен для добавления информации о расположении магазина при
отправке заказа.
3. Импортер (Importer) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Страна происхождения товара
(OriginCountry). Метод GetCompanyInfo() должен быть переопределен для
отображения страны происхождения товара вместе с остальной
информацией о компании.

```
using System;
using System.Collections.Generic;

public delegate void SupplierEventHandler(string message);

public static class SupplierEvents
{
    public static event SupplierEventHandler OnSupplierUpdated;
    public static event SupplierEventHandler OnOrderProcessed;

    public static void NotifyUpdate(string message) =>
        OnSupplierUpdated?.Invoke(message);

    public static void NotifyOrder(string message) =>
        OnOrderProcessed?.Invoke(message);
}

public interface IDiscountProvider
{
    double DiscountRate { get; set; }
    void ApplyDiscount(double percent);
}

public interface IInternational
{
    string Country { get; set; }
    void ShowInternationalStatus();
}

public class Supplier
{
    private int _supplierId;
    private string _companyName;
    private string _productType;

    public int SupplierId { get => _supplierId; set => _supplierId = value; }
    public string CompanyName { get => _companyName; set => _companyName = value; }
    public string ProductType { get => _productType; set => _productType = value; }

    public string CEO { get; set; }
    public DateTime RegistrationDate { get; set; }
    public string Phone { get; set; }
    public string Country { get; set; }
    public double Rating { get; set; }
    public string Email { get; set; }
    public bool IsActive { get; set; }

    public int CompletedOrders { get; set; }
    public double TotalRevenue { get; set; }
    public string Website { get; set; }
    public List<string> Partners { get; set; } = new();

    public Supplier(int supplierId, string companyName, string productType)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;

        Country = "Россия";
        Rating = 4.5;
        Email = $"{companyName.Replace(" ", "").ToLower()}@mail.com";
        Phone = "+7(900)000-00-00";
        CEO = "Директор не указан";
        RegistrationDate = DateTime.Now.AddYears(-3);
        IsActive = true;

        CompletedOrders = 0;
        TotalRevenue = 0;
        Website = "www.company.ru";
    }

    public void AddPartner(string partner)
    {
        Partners.Add(partner);
        SupplierEvents.NotifyUpdate($"{CompanyName} добавил партнёра: {partner}");
    }

    public void CompleteOrder(double amount)
    {
        CompletedOrders++;
        TotalRevenue += amount;

        SupplierEvents.NotifyOrder($"{CompanyName} выполнил заказ на сумму {amount} руб.");
    }

    public void Deactivate()
    {
        IsActive = false;
        SupplierEvents.NotifyUpdate($"{CompanyName} деактивирован.");
    }

    public void PrintStatistics()
    {
        Console.WriteLine($"{CompanyName}: выполнено заказов: {CompletedOrders}, доход: {TotalRevenue} руб.");
    }

    public virtual void GetCompanyInfo()
    {
        Console.WriteLine($"ID: {SupplierId}");
        Console.WriteLine($"Компания: {CompanyName}");
        Console.WriteLine($"Тип продукции: {ProductType}");
        Console.WriteLine($"Страна: {Country}");
        Console.WriteLine($"Рейтинг: {Rating}");
        Console.WriteLine($"Email: {Email}");
        Console.WriteLine($"Телефон: {Phone}");
        Console.WriteLine($"Гендиректор: {CEO}");
    }

    public virtual void SubmitOrder() =>
        Console.WriteLine($"Заказ отправлен поставщику {CompanyName}");
}

public class Manufacturer : Supplier
{
    public int FoundedYear { get; set; }
    public int ProductionCapacity { get; set; }
    public bool HasISO9001 { get; set; }
    public string MainProduct { get; set; }
    public int EmployeesCount { get; set; }
    public string FactoryAddress { get; set; }
    public bool HasAutomation { get; set; }
    public string EquipmentVendor { get; set; }

    public double EnergyConsumption { get; set; }
    public int MachinesCount { get; set; }
    public bool Works24h { get; set; }
    public List<string> Materials { get; set; } = new();

    public Manufacturer(int id, string name, string product, int foundedYear)
        : base(id, name, product)
    {
        FoundedYear = foundedYear;
        ProductionCapacity = 10000;
        HasISO9001 = true;
        MainProduct = product;
        EmployeesCount = 300;

        FactoryAddress = "г. Москва, индустриальный пр., 25";
        HasAutomation = true;
        EquipmentVendor = "Siemens";

        EnergyConsumption = 55000;
        MachinesCount = 120;
        Works24h = true;
    }

    public void UpgradeEquipment()
    {
        MachinesCount += 5;
        SupplierEvents.NotifyUpdate($"{CompanyName} обновил оборудование. Всего машин: {MachinesCount}");
    }

    public void AddMaterial(string material)
    {
        Materials.Add(material);
        Console.WriteLine($"{CompanyName}: добавлен материал: {material}");
    }

    public void CalculateEcoRating()
    {
        double rating = 100 - (EnergyConsumption / 1000);
        Console.WriteLine($"{CompanyName}: эко-рейтинг = {rating:F1}");
    }

    public void PauseProduction()
    {
        Console.WriteLine($"{CompanyName}: производство временно остановлено!");
    }
}

public class Retailer : Supplier, IDiscountProvider
{
    public string StoreLocation { get; set; }
    public int BranchCount { get; set; }
    public string ManagerName { get; set; }
    public double AverageMonthlySales { get; set; }
    public double DiscountRate { get; set; }
    public bool OnlineStore { get; set; }
    public int DailyCustomers { get; set; }
    public string WorkingHours { get; set; }

    public int ReturnsCount { get; set; }
    public double AverageCheck { get; set; }
    public List<string> Promotions { get; set; } = new();
    public bool HasSelfCheckout { get; set; }

    public Retailer(int id, string name, string product, string location)
        : base(id, name, product)
    {
        StoreLocation = location;
        BranchCount = 5;
        ManagerName = "Иванов И.И.";
        AverageMonthlySales = 1500000;
        DiscountRate = 5.0;

        OnlineStore = true;
        DailyCustomers = 600;
        WorkingHours = "09:00–21:00";

        ReturnsCount = 0;
        AverageCheck = 2500;
        HasSelfCheckout = false;
    }

    public void AddPromotion(string promo)
    {
        Promotions.Add(promo);
        Console.WriteLine($"{CompanyName}: новая акция — {promo}");
    }

    public void ProcessReturn()
    {
        ReturnsCount++;
        Console.WriteLine($"{CompanyName}: возврат обработан. Всего возвратов: {ReturnsCount}");
    }

    public void CalculateDailyRevenue()
    {
        Console.WriteLine($"{CompanyName}: дневная выручка ≈ {AverageCheck * DailyCustomers} руб.");
    }

    public void EnableSelfCheckout()
    {
        HasSelfCheckout = true;
        Console.WriteLine($"{CompanyName}: терминалы самообслуживания включены.");
    }

    void IDiscountProvider.ApplyDiscount(double percent)
    {
        DiscountRate = percent;
        Console.WriteLine($"(ЯВНО) {CompanyName} теперь даёт скидку {DiscountRate}%");
    }
}

public class Importer : Supplier, IInternational, IDiscountProvider
{
    public string OriginCountry { get; set; }
    public string LicenseNumber { get; set; }
    public double CustomsFee { get; set; }
    public string TransportType { get; set; }

    public double DiscountRate { get; set; }
    public string Country { get; set; }

    public bool HasBroker { get; set; }
    public string WarehouseLocation { get; set; }
    public int AnnualImports { get; set; }

    public double DeliveryTime { get; set; }
    public bool RequiresCooling { get; set; }
    public int ContainersCount { get; set; }
    public List<string> Routes { get; set; } = new();

    public Importer(int id, string name, string product, string origin)
        : base(id, name, product)
    {
        OriginCountry = origin;
        LicenseNumber = $"LIC-{id}-X";
        CustomsFee = 12.5;
        TransportType = "Морской транспорт";
        Country = origin;
        DiscountRate = 0;

        HasBroker = true;
        WarehouseLocation = "Склад №12, порт";
        AnnualImports = 120;

        DeliveryTime = 28;
        RequiresCooling = false;
        ContainersCount = 40;
    }

    public void AddRoute(string route)
    {
        Routes.Add(route);
        Console.WriteLine($"{CompanyName}: добавлен маршрут — {route}");
    }

    public void RequestBroker()
    {
        Console.WriteLine($"{CompanyName}: отправлен запрос брокеру.");
    }

    public void CalculateDeliveryTime()
    {
        Console.WriteLine($"{CompanyName}: доставка занимает {DeliveryTime} дней.");
    }

    public void UpdateContainers(int count)
    {
        ContainersCount += count;
        Console.WriteLine($"{CompanyName}: контейнеров теперь {ContainersCount}");
    }

    public void ShowInternationalStatus() =>
        Console.WriteLine($"{CompanyName} — международный поставщик из {OriginCountry}");

    public void ApplyDiscount(double percent)
    {
        DiscountRate = percent;
        Console.WriteLine($"{CompanyName} даёт скидку {DiscountRate}% на импорт");
    }
}

public class SupplierCollection<T> where T : Supplier
{
    private List<T> suppliers = new List<T>();

    public void Add(T supplier)
    {
        suppliers.Add(supplier);
    }

    public void DisplayAll()
    {
        Console.WriteLine("\n=== ВСЕ ПОСТАВЩИКИ ===");
        foreach (var supplier in suppliers)
        {
            supplier.GetCompanyInfo();
            Console.WriteLine("---");
        }
    }
}


SupplierEvents.OnSupplierUpdated += msg =>
{
    Console.WriteLine($"[EVENT: UPDATE] {msg}");
};

SupplierEvents.OnOrderProcessed += msg =>
{
    Console.WriteLine($"[EVENT: ORDER] {msg}");
};

Supplier supplier1 = new Supplier(1, "ООО 'Поставщик+'", "Электроника");
Manufacturer manufacturer1 = new Manufacturer(2, "Завод 'Электроникс'", "Компьютеры", 1995);
Retailer retailer1 = new Retailer(3, "Сеть 'ТехноМир'", "Бытовая техника", "ул. Центральная, 123");
Importer importer1 = new Importer(4, "ИмпортТрейд", "Мебель", "Италия");

supplier1.AddPartner("СибГрупп");
supplier1.CompleteOrder(150000);

manufacturer1.UpgradeEquipment();
manufacturer1.AddMaterial("Сталь");

retailer1.AddPromotion("Скидка 25%");
retailer1.EnableSelfCheckout();

importer1.AddRoute("Марсель — Санкт-Петербург");
importer1.UpdateContainers(10);

SupplierCollection<Supplier> collection = new SupplierCollection<Supplier>();
collection.Add(supplier1);
collection.Add(manufacturer1);
collection.Add(retailer1);
collection.Add(importer1);

collection.DisplayAll();

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ИНТЕРФЕЙСОВ ===");
        
IDiscountProvider discountRetailer = retailer1;
discountRetailer.ApplyDiscount(15.0);
        
IInternational internationalImporter = importer1;
internationalImporter.ShowInternationalStatus();
        
IDiscountProvider discountImporter = importer1;
    discountImporter.ApplyDiscount(7.5);


```
